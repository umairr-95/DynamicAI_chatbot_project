```python
# Step 2: Dataset Audit

## Objective

# The objective of this notebook is to audit the three approved datasets.

# We will:
# 1. Load the telecom Excel dataset.
# 2. Load the banking CSV dataset.
# 3. Load the second telecom CSV dataset.
# 4. Inspect dataset dimensions and column names.
# 5. Identify missing values and duplicate records.
# 6. Identify customer ID and churn target columns.
# 7. Compare customer IDs between the two telecom datasets.
# 8. Save an audit report.

# Banking and telecom datasets will remain separate.
```

```python
import pandas as pd
import numpy as np
import json
import re
from pathlib import Path

import warnings
warnings.filterwarnings("ignore")

print("Libraries imported successfully.")
```

```python
CURRENT_DIR = Path.cwd()

if CURRENT_DIR.name == "notebooks":
    PROJECT_ROOT = CURRENT_DIR.parent
else:
    PROJECT_ROOT = CURRENT_DIR

DATA_RAW = PROJECT_ROOT / "data" / "raw"
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"
REPORTS = PROJECT_ROOT / "outputs" / "evaluation_reports"

REPORTS.mkdir(parents=True, exist_ok=True)

# Approved dataset paths
telecom_excel_path = DATA_RAW / "D:\\Dynamic_AI_Chatbot_Project\\Raw_datsets\\Telco_customer_churn(1).xlsx"

banking_csv_path = DATA_RAW / "D:\\Dynamic_AI_Chatbot_Project\\Raw_datsets\\Customer-Churn-Records(1).csv"

telecom_csv_path = DATA_RAW / "D:\\Dynamic_AI_Chatbot_Project\\Raw_datsets\\Customer%20Churn.csv"

print("Project root:", PROJECT_ROOT)
print("Raw data folder:", DATA_RAW)

print("\nDataset paths:")
print(telecom_excel_path)
print(banking_csv_path)
print(telecom_csv_path)
```

```python
# Check all three files.

dataset_paths = {
    "Main Telecom Excel": telecom_excel_path,
    "Banking CSV": banking_csv_path,
    "Second Telecom CSV": telecom_csv_path
}

print("DATASET FILE CHECK")
print("-" * 60)

for dataset_name, file_path in dataset_paths.items():

    if file_path.exists():
        print(f"FOUND: {dataset_name}")
        print("File size:", round(file_path.stat().st_size / 1024, 2), "KB")
    else:
        print(f"MISSING: {dataset_name}")
        print("Expected location:", file_path)
```

```python
# Load the datasets.

def read_csv_safely(file_path):

    encodings = ["utf-8", "utf-8-sig", "cp1252", "latin1"]

    for encoding in encodings:
        try:
            return pd.read_csv(file_path, encoding=encoding)
        except UnicodeDecodeError:
            continue

    raise ValueError(f"Could not read CSV file: {file_path}")


# Main telecom dataset
telecom_excel = pd.read_excel(
    telecom_excel_path,
    engine="openpyxl"
)

# Banking dataset
banking_data = read_csv_safely(banking_csv_path)

# Second telecom dataset
telecom_csv = read_csv_safely(telecom_csv_path)


print("All datasets loaded successfully.")

print("\nMain Telecom Excel:", telecom_excel.shape)
print("Banking CSV:", banking_data.shape)
print("Second Telecom CSV:", telecom_csv.shape)
```

```python
# Display the first five records
print("MAIN TELECOM DATASET")
display(telecom_excel.head())

print("\nBANKING DATASET")
display(banking_data.head())

print("\nSECOND TELECOM DATASET")
display(telecom_csv.head())
```

```python
# Inspect column names and data types
datasets = {
    "Main Telecom Excel": telecom_excel,
    "Banking CSV": banking_data,
    "Second Telecom CSV": telecom_csv
}

for name, df in datasets.items():

    print("\n" + "=" * 65)
    print(name)
    print("=" * 65)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData types:")
    print(df.dtypes)

    print("\nDataset shape:")
    print(df.shape)
```

```python
# check missing values

for name, df in datasets.items():

    print("\n" + "=" * 65)
    print("MISSING VALUE REPORT:", name)
    print("=" * 65)

    missing = pd.DataFrame({
        "Missing_Count": df.isnull().sum(),
        "Missing_Percentage": (
            df.isnull().sum() / len(df) * 100
        ).round(2)
    })

    missing = missing.sort_values(
        by="Missing_Count",
        ascending=False
    )

    display(missing)
```

```python
# Check duplicate records
for name, df in datasets.items():

    print("\n" + "=" * 65)
    print("DUPLICATE REPORT:", name)
    print("=" * 65)

    print("Total records:", len(df))
    print("Duplicate rows:", df.duplicated().sum())
    print(
        "Unique rows:",
        len(df.drop_duplicates())
    )
```

```python
# Identify customer ID columns

def find_customer_id_column(df):

    possible_names = [
        "customerid",
        "customer_id",
        "customer id",
        "customer",
        "clientid",
        "client_id",
        "client id"
    ]

    normalized_columns = {
        re.sub(r"[^a-z0-9]", "", str(col).lower()): col
        for col in df.columns
    }

    for name in possible_names:

        normalized_name = re.sub(
            r"[^a-z0-9]", "", name.lower()
        )

        if normalized_name in normalized_columns:
            return normalized_columns[normalized_name]

    return None


for name, df in datasets.items():

    id_column = find_customer_id_column(df)

    print(name)
    print("Customer ID column:", id_column)
    print("-" * 50)
```

```python
# Compare the two telecom datasets

excel_id_column = find_customer_id_column(telecom_excel)
csv_id_column = find_customer_id_column(telecom_csv)

if excel_id_column is None or csv_id_column is None:

    print("Could not automatically identify both customer ID columns.")
    print("Excel ID column:", excel_id_column)
    print("CSV ID column:", csv_id_column)
    print("We must inspect the column names and identify them manually.")

else:

    excel_ids = set(
        telecom_excel[excel_id_column]
        .dropna()
        .astype(str)
        .str.strip()
    )

    csv_ids = set(
        telecom_csv[csv_id_column]
        .dropna()
        .astype(str)
        .str.strip()
    )

    overlapping_ids = excel_ids.intersection(csv_ids)

    print("TELECOM CUSTOMER ID COMPARISON")
    print("-" * 50)

    print("Unique Excel customer IDs:", len(excel_ids))
    print("Unique CSV customer IDs:", len(csv_ids))
    print("Overlapping customer IDs:", len(overlapping_ids))

    print(
        "Excel-only customer IDs:",
        len(excel_ids - csv_ids)
    )

    print(
        "CSV-only customer IDs:",
        len(csv_ids - excel_ids)
    )

    print("\nFirst 10 overlapping IDs:")
    print(list(overlapping_ids)[:10])
```

```python
# Identify possible churn target columns

possible_target_names = [
    "churn",
    "exited",
    "churnvalue",
    "churn_label",
    "attrition",
    "attrition_flag"
]

def find_target_column(df):

    normalized_columns = {
        re.sub(r"[^a-z0-9]", "", str(col).lower()): col
        for col in df.columns
    }

    for target in possible_target_names:

        normalized_target = re.sub(
            r"[^a-z0-9]", "", target.lower()
        )

        if normalized_target in normalized_columns:
            return normalized_columns[normalized_target]

    return None


for name, df in datasets.items():

    target_column = find_target_column(df)

    print("\nDataset:", name)
    print("Possible target column:", target_column)

    if target_column is not None:
        print("\nTarget values:")
        display(df[target_column].value_counts(dropna=False))
```

```python
# Generate an audit report

audit_report = {}

for name, df in datasets.items():

    audit_report[name] = {
        "rows": int(df.shape[0]),
        "columns_count": int(df.shape[1]),
        "column_names": [str(col) for col in df.columns],
        "duplicate_rows": int(df.duplicated().sum()),
        "missing_values": {
            str(col): int(count)
            for col, count in df.isnull().sum().items()
        },
        "customer_id_column": find_customer_id_column(df),
        "possible_target_column": find_target_column(df)
    }


if excel_id_column is not None and csv_id_column is not None:

    audit_report["Telecom_ID_Comparison"] = {
        "excel_unique_ids": len(excel_ids),
        "csv_unique_ids": len(csv_ids),
        "overlapping_ids": len(overlapping_ids)
    }


report_path = REPORTS / "dataset_audit_report.json"

with open(report_path, "w", encoding="utf-8") as file:
    json.dump(audit_report, file, indent=4)


print("Audit report saved successfully:")
print(report_path)
```

```python
# Final audit summary

print("=" * 65)
print("DATASET AUDIT COMPLETED")
print("=" * 65)

for name, df in datasets.items():

    print("\nDataset:", name)
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Duplicate rows:", df.duplicated().sum())
    print("Customer ID:", find_customer_id_column(df))
    print("Churn target:", find_target_column(df))

print("\nAudit report location:", report_path)
```
