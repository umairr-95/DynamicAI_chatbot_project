```python
# Check project files

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path(r"D:\Dynamic_AI_Chatbot_Project")

required_files = [
    "Notebooks/01_Project_Setup.ipynb",
    "Notebooks/Dataset_Audit.ipynb",
    "Notebooks/Data Cleaning and Preprocessing.ipynb",
    "data/processed/telecom_churn_cleaned.xlsx",
    "data/processed/banking_churn_cleaned.csv",
    "data/processed/telecom_churn_second_cleaned.csv",
    "data/model_ready/telecom_excel_train.csv",
    "data/model_ready/telecom_excel_test.csv",
    "data/model_ready/telecom_csv_train.csv",
    "data/model_ready/telecom_csv_test.csv",
    "data/model_ready/banking_train.csv",
    "data/model_ready/banking_test.csv",
    "models/nlp/intent_logistic_regression.pkl",
    "outputs/sentiment/sentiment_emotion_report.csv",
    "outputs/self_learning/interaction_history.csv",
    "outputs/self_learning/user_feedback.csv",
    "outputs/analytics/chatbot_kpi_summary.csv",
    "outputs/chatbot/chatbot_interaction_history.csv",
    "outputs/chatbot/final_chatbot_test_report.csv"
]

validation_results = []

for relative_path in required_files:
    file_path = PROJECT_ROOT / relative_path

    validation_results.append({
        "File": relative_path,
        "Exists": file_path.exists()
    })

validation_df = pd.DataFrame(validation_results)

display(validation_df)

print("Files found:", validation_df["Exists"].sum())
print("Files missing:", (~validation_df["Exists"]).sum())
```

```python
from pathlib import Path

PROJECT_ROOT = Path(r"D:\Dynamic_AI_Chatbot_Project")
MODELS_DIR = PROJECT_ROOT / "models"

print("ALL SAVED MODEL FILES")
print("=" * 50)

model_files = []

for extension in ["*.pkl", "*.joblib"]:
    model_files.extend(MODELS_DIR.rglob(extension))

if model_files:
    for file_path in sorted(model_files):
        print(file_path.relative_to(PROJECT_ROOT))
else:
    print("No saved model files found.")
```

```python
from pathlib import Path
import joblib

PROJECT_ROOT = Path(r"D:\Dynamic_AI_Chatbot_Project")

model_files = [
    "models/telecom_excel_best_churn_model.joblib",
    "models/telecom_csv_best_churn_model.joblib",
    "models/banking_best_churn_model.joblib",
    "models/nlp/intent_logistic_regression.pkl"
]

print("FINAL MODEL VALIDATION")
print("=" * 55)

validation_results = []

for relative_path in model_files:
    model_path = PROJECT_ROOT / relative_path

    try:
        model = joblib.load(model_path)

        validation_results.append({
            "Model": relative_path,
            "Exists": True,
            "Loads Successfully": True,
            "Model Type": type(model).__name__
        })

    except Exception as error:
        validation_results.append({
            "Model": relative_path,
            "Exists": model_path.exists(),
            "Loads Successfully": False,
            "Model Type": str(error)
        })

validation_df = pd.DataFrame(validation_results)

display(validation_df)

print("Models validated:", len(validation_df))
print(
    "Successfully loaded:",
    validation_df["Loads Successfully"].sum()
)
```
