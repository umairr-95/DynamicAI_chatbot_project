```python
# Step 8: Sentiment Analysis and Emotion Detection

## Objective
# This notebook implements sentiment analysis and basic emotion detection
# for the Dynamic AI Chatbot.

# ## Key Tasks
# 1. Identify positive, negative, and neutral sentiment.
# 2. Detect basic emotional signals.
# 3. Generate sentiment-aware response suggestions.
# 4. Test the system using sample conversational messages.
# 5. Save sentiment analysis results.

# ## Tools
# - Python
# - VADER Sentiment Analysis
# - Pandas
# - Regular Expressions
```

```python
import pandas as pd
import re
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
from pathlib import Path

print("Sentiment analysis libraries imported successfully.")
```

```python
# Initialize VADER sentiment analyzer

analyzer = SentimentIntensityAnalyzer()

def analyze_sentiment(message):

    scores = analyzer.polarity_scores(message)
    compound_score = scores["compound"]

    if compound_score >= 0.05:
        sentiment = "Positive"
    elif compound_score <= -0.05:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    return {
        "message": message,
        "sentiment": sentiment,
        "positive_score": scores["pos"],
        "negative_score": scores["neg"],
        "neutral_score": scores["neu"],
        "compound_score": compound_score
    }

print("Sentiment analysis function created successfully.")
```

```python
# Test different messages

test_messages = [
    "Thank you! Your service is excellent.",
    "I am very unhappy with your customer support.",
    "I want to know your business hours.",
    "I am frustrated because my internet is not working.",
    "Everything is working perfectly. I am very happy."
]

sentiment_results = []

for message in test_messages:
    result = analyze_sentiment(message)
    sentiment_results.append(result)

sentiment_df = pd.DataFrame(sentiment_results)

display(sentiment_df)
```

```python
# Create the emotion detection function

emotion_keywords = {
    "Happy": [
        "happy", "excited", "wonderful", "great",
        "excellent", "pleased", "delighted"
    ],
    "Angry": [
        "angry", "furious", "annoyed", "hate",
        "unacceptable", "irritated"
    ],
    "Sad": [
        "sad", "upset", "disappointed", "unhappy",
        "depressed", "sorry"
    ],
    "Frustrated": [
        "frustrated", "problem", "issue", "failed",
        "not working", "fed up", "trouble"
    ],
    "Worried": [
        "worried", "concerned", "afraid", "anxious",
        "scared", "uncertain"
    ]
}

def detect_emotion(message):

    text = message.lower()
    detected_emotions = []

    for emotion, keywords in emotion_keywords.items():
        for keyword in keywords:
            if keyword in text:
                detected_emotions.append(emotion)
                break

    if not detected_emotions:
        detected_emotions = ["No specific emotion detected"]

    return detected_emotions

print("Emotion detection function created successfully.")
```

```python
# Run emotion tests

emotion_test_messages = [
    "I am very happy with your service.",
    "I am angry because my internet is not working.",
    "I feel disappointed with the support team.",
    "I am worried about my account.",
    "Please tell me your service timings."
]

for message in emotion_test_messages:
    emotions = detect_emotion(message)

    print("Message:", message)
    print("Detected emotion:", emotions)
    print("-" * 60)
```

```python
# Response generation

def generate_adaptive_response(message):

    sentiment_result = analyze_sentiment(message)
    emotions = detect_emotion(message)

    sentiment = sentiment_result["sentiment"]

    if "Angry" in emotions or "Frustrated" in emotions:
        response = (
            "I understand your frustration. "
            "I'm sorry you're experiencing this issue. "
            "Let me help you find a solution."
        )

    elif "Sad" in emotions or "Worried" in emotions:
        response = (
            "I understand your concern. "
            "Don't worry, I'll try my best to assist you."
        )

    elif sentiment == "Positive":
        response = (
            "I'm glad to hear that! "
            "Thank you for your positive feedback."
        )

    elif sentiment == "Negative":
        response = (
            "I'm sorry to hear that. "
            "Could you please explain the issue so I can help?"
        )

    else:
        response = (
            "Thank you for your message. "
            "How can I assist you further?"
        )

    return {
        "sentiment": sentiment,
        "emotions": emotions,
        "response": response
    }


print("Adaptive response function created successfully.")
```

```python
# Test different user messages

adaptive_test_messages = [
    "I am very happy with your service.",
    "I am angry because my internet is not working.",
    "I am worried about my account.",
    "Your customer support is terrible.",
    "Can you tell me about your services?"
]

for message in adaptive_test_messages:

    result = generate_adaptive_response(message)

    print("USER:", message)
    print("SENTIMENT:", result["sentiment"])
    print("EMOTION:", result["emotions"])
    print("CHATBOT:", result["response"])
    print("-" * 70)
```

```python
# Generate analysis results

analysis_results = []

for message in adaptive_test_messages:

    sentiment_result = analyze_sentiment(message)
    emotions = detect_emotion(message)
    adaptive_result = generate_adaptive_response(message)

    analysis_results.append({
        "Message": message,
        "Sentiment": sentiment_result["sentiment"],
        "Positive Score": sentiment_result["positive_score"],
        "Negative Score": sentiment_result["negative_score"],
        "Neutral Score": sentiment_result["neutral_score"],
        "Compound Score": sentiment_result["compound_score"],
        "Detected Emotions": ", ".join(emotions),
        "Chatbot Response": adaptive_result["response"]
    })

sentiment_report = pd.DataFrame(analysis_results)

display(sentiment_report)
```

```python
# Save results

PROJECT_ROOT = Path(r"D:\Dynamic_AI_Chatbot_Project")

OUTPUT_DIR = PROJECT_ROOT / "outputs" / "sentiment"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

report_path = OUTPUT_DIR / "sentiment_emotion_report.csv"

sentiment_report.to_csv(report_path, index=False)

print("Sentiment and emotion report saved successfully.")
print("Saved location:", report_path)
```
